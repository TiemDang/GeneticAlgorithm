W = np.random.rand(2,5)* 2 - 1

```python
def vd_nn(X,W,V):
    def af(x):
        y = np.tanh(x)
        return y
    net_h = W.T @ X  # --> Size (3,5) transpose to (5x3), (5x3) x (3x1) = Size 5x1
    y_h = af(net_h)
    output = V.T @ y_h # --> Size 2x1
    return output

y = vd_nn(X,W,V)
print(y)
```
